Panda

```python
import pandas as pd
```

```python
#Creating a series
s = pd.Series([1,2,3,4,5], name = 'numbers')
print(s)
```

```python
 #Creating a Data frame
df = pd.DataFrame( {'A':[ 1, 2, 3], 'B':['a', 'b', 'c'], 'C':[4.5, 5.5, 6.5] } )
print(df)
```

Data Manipulation

```python
#Accessiing a specific value
value = df.loc[ 0, 'C']
print(value)

# Selecting multiple columns
selected_columns = df.loc[:, ['A', 'C']]
print("\n",selected_columns)

# Selecting a range of rows
selected_rows = df.loc[ 2:3, :]
print("\n",selected_rows)

#Selecting both
scol = df.loc[ 0:1, ['A', 'C']]
print("\n",scol)
```

```python
#Filtering rows based on condition
filtered_df = df[ df['A'] > 2]
print(filtered_df)

#Filtering rows based on multiple conditions
filtered_df = df[ ( df['A'] > 1 ) & ( df['C'] < 6 ) ]
print("\n",filtered_df)
```

Data cleaning

```python
import pandas as pd
import numpy as np

#Creating a DataFrame with missing values
df = pd.DataFrame( {
    'A':[1, 2, np.nan, 4],
    'B':[np.nan, 'b', 'c', 'd'],
    'C': [4.5, np.nan, 6.5, 7.5]
} )

#Detecting missing values
missing_values = df.isna()
print(missing_values)
```

```python
#Filling missing data with a specific value
filled_df = df.fillna(-1)
print(filled_df)
```

Transforming a dataframe

```python
#Creating a DataFrame for traansformation example
df = pd.DataFrame( {
    'A':[1, 2, 3, 4], 
    'B':[10, 20, 30, 40]
} )

#Applying a transformation function
transformed_df = df.apply(lambda x: x * 4)
print(transformed_df)
```
