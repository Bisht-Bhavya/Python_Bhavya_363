Palindrome

```python
n = int(input("Enter a number: "))
temp = n
sum = 0
while n > 0:
    a = n % 10
    sum = (sum * 10) + a
    n = n // 10
if sum == temp:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
import math
def perfect_square(x):
    sqr = math.sqrt(x)
    if sqr ** 2 == x:
        return 1
    else:
        return -1

print(perfect_square(5))
```

```python
def removenth(str, n):
    if n > len(str):
        return str
    temp = ""
    for x in str:
        if x == str[n]:
            continue
        else:
            temp +=x
    return temp

print(removenth('Bhavya',0))
```

```python
str = input('Enter a sequence of words').split(',')
n = len(str)
for x in range(n):
    str[x] = str[x].strip()
for i in range(n):
    for j in range(0, n - i - 1):
        if str[j] > str[j+1]:
             str[j+1], str[j] = str[j], str[j+1]

print(','.join(str))
```

```python
passwords = input("Enter comma-separated passwords: ")
password_list = passwords.split(",")

valid_passwords = []

for pwd in password_list:
    pwd = pwd.strip()
    
    if 6 <= len(pwd) <= 12:
        has_lower = False
        has_upper = False
        has_digit = False
        has_special = False
        
        for ch in pwd:
            if 'a' <= ch <= 'z':
                has_lower = True
            elif 'A' <= ch <= 'Z':
                has_upper = True
            elif '0' <= ch <= '9':
                has_digit = True
            elif ch in "$#@":
                has_special = True
        
        if has_lower and has_upper and has_digit and has_special:
            valid_passwords.append(pwd)

print(",".join(valid_passwords))
```
