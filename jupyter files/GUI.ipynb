```python
#Sample data
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 25, 30, 35])
```

```python
# Create a plot line
plt.plot( x ,y , marker = '.', color = 'g')

#Add Title and labels
plt.title('Basic Line Plot')
plt.xlabel(' X-axis Labal')
plt.ylabel(' Y-axis Labal')

#Show Grid
plt.grid(True)

#Display the plot
plt.show()
```

Histogram

```python
import matplotlib.pyplot as plt
```

```python
#Sample Data
categories = [ 'A', 'B', 'C', 'D']
values = [ 10, 20, 15, 25]
```

```python
#Create a bar plot
plt.bar(categories, values, color = 'skyblue')

#Add title and labels
plt.title('Basic Bar plot')
plt.xlabel('Categories')
plt.ylabel('Values')

#Show Grid
plt.grid( axis = 'y', linestyle = '-', alpha = 0.7)

#Display the plot
plt.show()
```

Horizontal Bar

```python
plt.barh(categories, values, color='skyblue')
```

Pie Chart

```python
#Sample data
labels = ['A', 'B', 'C', 'D']
sizes = [ 15, 30, 45, 10]

#Create a pie chart
plt.pie(sizes, labels = labels)

plt.show()
```

Area Plot

```python
#Sample data
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

#Create an area plot
plt.fill_between( x, y, color="skyblue", alpha = 0.4)

#Add title and labels
plt.title('Basic Area Plot')
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')

plt.show()
```

GUI - Tkinter

```python
# Creation of Button
import tkinter as tk

def on_button_click():
    print("Button Clicked!")

root = tk.Tk() # Creation of window
button = tk.Button(root, text = "Click Me", command = on_button_click) # Creates a button
button.pack() # The text will be inside the window
root.mainloop()
```

```python
#Creation of label
import tkinter as tk

root = tk.Tk()
label = tk.Label(root, text = "This is a label") # Creates a label with the text
label.pack() # Adds the label to the window
root.mainloop()
```

```python
# Creation of Entry
import tkinter as tk

root = tk.Tk()
entry = tk.Entry(root) # Creates a single line text input
entry.pack() # Adds the entry field to the window
root.mainloop()
```

```python
# Creation of text
import tkinter as tk

root = tk.Tk()
text = tk.Text(root) # Creates a multi line text input field
text.pack() # Adds the text field to the window
root.mainloop()
```

```python
# Creation of Checkbutton
import tkinter as tk

root = tk.Tk()
checkbox_var = tk.IntVar() # Creates an integer variable to hold the state of the checkbox
checkbox = tk.Checkbutton(root, text = "Check me", variable = checkbox_var)
checkbox.pack() # Adds the checkbox to the window
root.mainloop()
```

GUI Example

```python
import tkinter as tk

def update_label():
    name = entry.get()
    label.config( text = f"Hello, {name}!")

# Create the main window
root = tk.Tk()
root.title("Simple GUI Example")
root.geometry("300x200")

# Create and place a label widget
label = tk.Label( root,text = "Enter your name: ")
label.pack( pady = 10)

# Create and pack an entry widget
entry = tk.Entry( root )
entry.pack( pady = 5 )

# Create and pack a button widget
button = tk.Button( root, text = "Submit", command = update_label )
button.pack( pady = 10 )

# Run the main event loop
root.mainloop()
```

Simple Calculator

```python
import tkinter as tk

# Function definations
def add():
    result.set(  float ( entry1.get() ) + float( entry2.get() ) )

def subtract():
    result.set(  float ( entry1.get() )  - float( entry2.get() ) )

def multiply():
    result.set(  float ( entry1.get() ) * float( entry2.get() ) )

def divide():
    result.set(  float ( entry1.get() ) / float( entry2.get() ) )

# Create main window
root = tk.Tk()
root.title("Simple Calculator ")
root.geometry("500x500")

# Create Entry Widgets
entry1 = tk.Entry(root)
entry1.pack()
entry2 = tk.Entry(root)
entry2.pack()

# Create Variable for result
result = tk.DoubleVar()

# Create result label
result_label = tk.Label( root, textvariable = result )
result_label.pack()

# Create Button
add_button = tk.Button( root, text= "ADD", command = add)
add_button.pack()

subtract_button = tk.Button( root, text= "Subtract", command = subtract)
subtract_button.pack()

multiply_button = tk.Button( root, text= "Multiply", command = multiply)
multiply_button.pack()

divide_button = tk.Button( root, text= "divide", command = divide)
divide_button.pack()

# Run the main loop
root.mainloop()
```
